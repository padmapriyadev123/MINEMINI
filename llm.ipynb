```python
# MP-C18 Citation Intent Classifier. Run as a Colab notebook; cell markers are '# %%'.

# %% [markdown]
# # MP-C18: Citation Intent Classifier (GPT-2 small + SciCite)
# Run cells **top to bottom**. Runtime → Change runtime type → **T4 GPU** first.
# Note: `GPTModel` / `load_weights_into_gpt` below are a reconstruction compatible with the standard "LLM from scratch" structure. Replace with your course files if you have them.

# %%
# CELL 1: setup, seed, device
get_ipython().system('pip install -q tiktoken gradio')

import json, os, glob, random, time, copy
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import tiktoken
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, classification_report

SEED = 123
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
tokenizer = tiktoken.get_encoding("gpt2")
PAD_ID = 50256

def encode(text):
    # disallowed_special=() so user text like "<|endoftext|>" cannot crash or inject the pad token
    return tokenizer.encode(text, disallowed_special=())

print("Device:", device, "| Seed:", SEED)

# %%
# CELL 2: get the GPT-2 weight downloader (use your course's gpt_download.py if you have one)
get_ipython().system('wget -q -nc https://raw.githubusercontent.com/rasbt/LLMs-from-scratch/main/ch05/01_main-chapter-code/gpt_download.py')
print("gpt_download.py found:", os.path.exists("gpt_download.py"))

# %%
# CELL 3: GPTModel architecture (reconstructed, GPT-2 small layout)
GPT_CONFIG_124M = {
    "vocab_size": 50257, "context_length": 1024, "emb_dim": 768,
    "n_heads": 12, "n_layers": 12, "drop_rate": 0.0, "qkv_bias": True
}

class LayerNorm(nn.Module):
    def __init__(self, emb_dim):
        super().__init__()
        self.eps = 1e-5
        self.scale = nn.Parameter(torch.ones(emb_dim))
        self.shift = nn.Parameter(torch.zeros(emb_dim))
    def forward(self, x):
        mean = x.mean(dim=-1, keepdim=True)
        var = x.var(dim=-1, keepdim=True, unbiased=False)
        return self.scale * (x - mean) / torch.sqrt(var + self.eps) + self.shift

class GELU(nn.Module):
    def forward(self, x):
        return 0.5 * x * (1 + torch.tanh(torch.sqrt(torch.tensor(2.0 / torch.pi)) *
                                         (x + 0.044715 * x ** 3)))

class FeedForward(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        self.layers = nn.Sequential(
            nn.Linear(cfg["emb_dim"], 4 * cfg["emb_dim"]), GELU(),
            nn.Linear(4 * cfg["emb_dim"], cfg["emb_dim"]))
    def forward(self, x):
        return self.layers(x)

class MultiHeadAttention(nn.Module):
    def __init__(self, d_in, d_out, context_length, dropout, num_heads, qkv_bias=False):
        super().__init__()
        assert d_out % num_heads == 0
        self.d_out, self.num_heads = d_out, num_heads
        self.head_dim = d_out // num_heads
        self.W_query = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.W_key = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.W_value = nn.Linear(d_in, d_out, bias=qkv_bias)
        self.out_proj = nn.Linear(d_out, d_out)
        self.dropout = nn.Dropout(dropout)
        self.register_buffer("mask", torch.triu(torch.ones(context_length, context_length), diagonal=1))
    def forward(self, x):
        b, n, _ = x.shape
        q = self.W_query(x).view(b, n, self.num_heads, self.head_dim).transpose(1, 2)
        k = self.W_key(x).view(b, n, self.num_heads, self.head_dim).transpose(1, 2)
        v = self.W_value(x).view(b, n, self.num_heads, self.head_dim).transpose(1, 2)
        scores = q @ k.transpose(2, 3)
        scores = scores.masked_fill(self.mask.bool()[:n, :n], -torch.inf)
        w = self.dropout(torch.softmax(scores / self.head_dim ** 0.5, dim=-1))
        ctx = (w @ v).transpose(1, 2).contiguous().view(b, n, self.d_out)
        return self.out_proj(ctx)

class TransformerBlock(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        self.att = MultiHeadAttention(cfg["emb_dim"], cfg["emb_dim"], cfg["context_length"],
                                      cfg["drop_rate"], cfg["n_heads"], cfg["qkv_bias"])
        self.ff = FeedForward(cfg)
        self.norm1, self.norm2 = LayerNorm(cfg["emb_dim"]), LayerNorm(cfg["emb_dim"])
        self.drop_shortcut = nn.Dropout(cfg["drop_rate"])
    def forward(self, x):
        x = x + self.drop_shortcut(self.att(self.norm1(x)))
        x = x + self.drop_shortcut(self.ff(self.norm2(x)))
        return x

class GPTModel(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        self.tok_emb = nn.Embedding(cfg["vocab_size"], cfg["emb_dim"])
        self.pos_emb = nn.Embedding(cfg["context_length"], cfg["emb_dim"])
        self.drop_emb = nn.Dropout(cfg["drop_rate"])
        self.trf_blocks = nn.Sequential(*[TransformerBlock(cfg) for _ in range(cfg["n_layers"])])
        self.final_norm = LayerNorm(cfg["emb_dim"])
        self.out_head = nn.Linear(cfg["emb_dim"], cfg["vocab_size"], bias=False)
    def forward(self, idx):
        b, n = idx.shape
        x = self.tok_emb(idx) + self.pos_emb(torch.arange(n, device=idx.device))
        x = self.trf_blocks(self.drop_emb(x))
        return self.out_head(self.final_norm(x))

print("Architecture defined.")

# %%
# CELL 4: load_weights_into_gpt + load pretrained GPT-2 124M + sanity test
from gpt_download import download_and_load_gpt2

def assign(left, right):
    if left.shape != right.shape:
        raise ValueError(f"Shape mismatch. Left: {left.shape}, Right: {right.shape}")
    return torch.nn.Parameter(torch.tensor(right))

def load_weights_into_gpt(gpt, params):
    gpt.pos_emb.weight = assign(gpt.pos_emb.weight, params["wpe"])
    gpt.tok_emb.weight = assign(gpt.tok_emb.weight, params["wte"])
    for b in range(len(params["blocks"])):
        blk, p = gpt.trf_blocks[b], params["blocks"][b]
        qw, kw, vw = np.split(p["attn"]["c_attn"]["w"], 3, axis=-1)
        blk.att.W_query.weight = assign(blk.att.W_query.weight, qw.T)
        blk.att.W_key.weight = assign(blk.att.W_key.weight, kw.T)
        blk.att.W_value.weight = assign(blk.att.W_value.weight, vw.T)
        qb, kb, vb = np.split(p["attn"]["c_attn"]["b"], 3, axis=-1)
        blk.att.W_query.bias = assign(blk.att.W_query.bias, qb)
        blk.att.W_key.bias = assign(blk.att.W_key.bias, kb)
        blk.att.W_value.bias = assign(blk.att.W_value.bias, vb)
        blk.att.out_proj.weight = assign(blk.att.out_proj.weight, p["attn"]["c_proj"]["w"].T)
        blk.att.out_proj.bias = assign(blk.att.out_proj.bias, p["attn"]["c_proj"]["b"])
        blk.ff.layers[0].weight = assign(blk.ff.layers[0].weight, p["mlp"]["c_fc"]["w"].T)
        blk.ff.layers[0].bias = assign(blk.ff.layers[0].bias, p["mlp"]["c_fc"]["b"])
        blk.ff.layers[2].weight = assign(blk.ff.layers[2].weight, p["mlp"]["c_proj"]["w"].T)
        blk.ff.layers[2].bias = assign(blk.ff.layers[2].bias, p["mlp"]["c_proj"]["b"])
        blk.norm1.scale = assign(blk.norm1.scale, p["ln_1"]["g"])
        blk.norm1.shift = assign(blk.norm1.shift, p["ln_1"]["b"])
        blk.norm2.scale = assign(blk.norm2.scale, p["ln_2"]["g"])
        blk.norm2.shift = assign(blk.norm2.shift, p["ln_2"]["b"])
    gpt.final_norm.scale = assign(gpt.final_norm.scale, params["g"])
    gpt.final_norm.shift = assign(gpt.final_norm.shift, params["b"])
    gpt.out_head.weight = assign(gpt.out_head.weight, params["wte"])  # weight tying

settings, params = download_and_load_gpt2(model_size="124M", models_dir="gpt2")
model = GPTModel(GPT_CONFIG_124M)
load_weights_into_gpt(model, params)
model.to(device).eval()

# Sanity test: output must be readable English. Gibberish = weights loaded wrongly.
ids = torch.tensor(encode("Every effort moves you")).unsqueeze(0).to(device)
with torch.no_grad():
    for _ in range(15):
        nxt = model(ids)[:, -1, :].argmax(dim=-1, keepdim=True)
        ids = torch.cat([ids, nxt], dim=1)
print(tokenizer.decode(ids[0].tolist()))

# %%
# CELL 5: download SciCite and locate the files
get_ipython().system('wget -q -nc https://s3-us-west-2.amazonaws.com/ai2-s2-research/scicite/scicite.tar.gz')
get_ipython().system('tar -xzf scicite.tar.gz')

def find_file(name):
    hits = glob.glob(f"**/{name}", recursive=True)
    assert hits, f"{name} not found. Run: !ls -R | head -50 and send me the output"
    return hits[0]

TRAIN_PATH, VAL_PATH, TEST_PATH = find_file("train.jsonl"), find_file("dev.jsonl"), find_file("test.jsonl")
print(TRAIN_PATH, VAL_PATH, TEST_PATH)
with open(TRAIN_PATH) as f:
    first = json.loads(f.readline())
print({k: (str(v)[:80]) for k, v in first.items()})

# %%
# CELL 6: load + preprocess + dataset statistics (save these numbers for the report)
LABELS = ["background", "method", "result"]
label2id = {l: i for i, l in enumerate(LABELS)}

def load_split(path):
    rows, skipped = [], 0
    with open(path) as f:
        for line in f:
            d = json.loads(line)
            text = " ".join(d["string"].split()).strip()   # normalise whitespace
            lab = d["label"]
            lab = label2id.get(lab) if isinstance(lab, str) else int(lab)
            if text and lab is not None:
                rows.append((text, lab))
            else:
                skipped += 1
    return rows, skipped

train_rows, s1 = load_split(TRAIN_PATH)
val_rows, s2 = load_split(VAL_PATH)
test_rows, s3 = load_split(TEST_PATH)
print("Skipped (empty text / unknown label):", s1, s2, s3)

# Optional subset (allowed by the rules). Set to e.g. 4000 to speed up; None = use all.
SUBSET_TRAIN = None
if SUBSET_TRAIN:
    random.Random(SEED).shuffle(train_rows)
    train_rows = train_rows[:SUBSET_TRAIN]
    print("Using random subset of train:", SUBSET_TRAIN)

print("Sizes -> train/val/test:", len(train_rows), len(val_rows), len(test_rows))
for name, rows in [("train", train_rows), ("val", val_rows), ("test", test_rows)]:
    counts = np.bincount([r[1] for r in rows], minlength=3)
    print(f"{name:5s}", dict(zip(LABELS, counts.tolist())), "| majority share: %.3f" % (counts.max() / counts.sum()))

lens = [len(encode(t)) for t, _ in train_rows]
print("Token length (train): mean %.1f, median %d, 95th pct %d, max %d" %
      (np.mean(lens), np.median(lens), np.percentile(lens, 95), max(lens)))

# %%
# CELL 7: Dataset (truncate + pad with 50256) and DataLoaders
MAX_LEN = 128
BATCH_SIZE = 16   # if CUDA out-of-memory: set 8 and re-run from this cell

class CitationDataset(Dataset):
    def __init__(self, rows, max_length=MAX_LEN, pad_id=PAD_ID):
        self.labels = [r[1] for r in rows]
        encoded = [encode(r[0]) for r in rows]
        self.n_truncated = sum(len(e) > max_length for e in encoded)
        self.encoded = [e[:max_length] + [pad_id] * (max_length - len(e[:max_length])) for e in encoded]
    def __getitem__(self, i):
        return torch.tensor(self.encoded[i]), torch.tensor(self.labels[i])
    def __len__(self):
        return len(self.labels)

train_ds, val_ds, test_ds = CitationDataset(train_rows), CitationDataset(val_rows), CitationDataset(test_rows)
print("Truncated sentences (train/val/test):", train_ds.n_truncated, val_ds.n_truncated, test_ds.n_truncated)

g = torch.Generator().manual_seed(SEED)
train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, drop_last=True, generator=g)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False)

x, y = next(iter(train_loader))
print("Batch:", x.shape, y.shape)   # expect [16, 128] and [16]

# %%
# CELL 7B: figure for the report (Figure 2.1): class distribution + token-length histogram
counts_train = np.bincount([r[1] for r in train_rows], minlength=3)
fig, axes = plt.subplots(1, 2, figsize=(10, 3.8))
bars = axes[0].bar(LABELS, counts_train, color=["#4C72B0", "#55A868", "#C44E52"])
for b_, c_ in zip(bars, counts_train):
    axes[0].text(b_.get_x() + b_.get_width() / 2, c_, str(int(c_)), ha="center", va="bottom")
axes[0].set_title("Training class distribution"); axes[0].set_ylabel("Sentences")
axes[1].hist(lens, bins=40, color="#8172B2")
axes[1].axvline(MAX_LEN, color="red", linestyle="--", label=f"MAX_LEN = {MAX_LEN}")
axes[1].set_title("Token length (train)"); axes[1].set_xlabel("Tokens per sentence")
axes[1].legend()
plt.tight_layout(); plt.savefig("data_stats.png", dpi=200); plt.show()

# %%
# CELL 8: new 3-class head + unfreeze selected layers
N_UNFREEZE_BLOCKS = 2   # last N transformer blocks are trained (try 1 for faster, 3-4 for more capacity)

for p in model.parameters():
    p.requires_grad = False

torch.manual_seed(SEED)   # reproducible random head
model.out_head = nn.Linear(GPT_CONFIG_124M["emb_dim"], 3).to(device)   # new head: trainable by default

for blk in model.trf_blocks[-N_UNFREEZE_BLOCKS:]:
    for p in blk.parameters():
        p.requires_grad = True
for p in model.final_norm.parameters():
    p.requires_grad = True

total = sum(p.numel() for p in model.parameters())
trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total params: {total:,} | Trainable: {trainable:,} ({100 * trainable / total:.1f}%)")

# %%
# CELL 9: helpers + BASELINE (before fine-tuning)
def get_logits(model, x):
    """Class logits taken at the last REAL token (padding is ignored).
    GPT-2 is causal, so the last real token has seen the whole sentence."""
    out = model(x)                                        # [batch, seq, 3]
    last = (x != PAD_ID).sum(dim=1).clamp(min=1) - 1      # index of last real token
    return out[torch.arange(x.size(0), device=x.device), last]

def calc_loss_batch(x, y, model, device):
    x, y = x.to(device), y.to(device)
    return F.cross_entropy(get_logits(model, x), y)

@torch.no_grad()
def evaluate_loader(loader, model, device, max_batches=None):
    model.eval()
    losses, preds, labels = [], [], []
    for i, (x, y) in enumerate(loader):
        if max_batches and i >= max_batches:
            break
        x, y = x.to(device), y.to(device)
        logits = get_logits(model, x)
        losses.append(F.cross_entropy(logits, y).item())
        preds += logits.argmax(-1).tolist()
        labels += y.tolist()
    return float(np.mean(losses)), np.array(preds), np.array(labels)

def metrics(preds, labels):
    return {"acc": round(accuracy_score(labels, preds), 4),
            "macro_f1": round(f1_score(labels, preds, average="macro", zero_division=0), 4)}

b_loss, b_pred, b_lab = evaluate_loader(test_loader, model, device)
baseline = metrics(b_pred, b_lab)
print("BASELINE test (random head, before training):", baseline, "| loss: %.4f" % b_loss)
print("Predicted class counts:", np.bincount(b_pred, minlength=3).tolist(), "| true:", np.bincount(b_lab, minlength=3).tolist())
print("Majority-class accuracy on test: %.4f" % (np.bincount(b_lab).max() / len(b_lab)))

# %%
# CELL 10: training loop (keeps the best-validation-loss weights)
# NOTE: always re-run CELL 8 and CELL 9 before re-training, otherwise you continue from the trained model.
EPOCHS = 4
LR = 5e-5
EVAL_FREQ = 100

optimizer = torch.optim.AdamW([p for p in model.parameters() if p.requires_grad], lr=LR, weight_decay=0.1)

steps, train_losses, val_losses = [], [], []
epoch_log = []
best = {"val_loss": float("inf"), "state": None, "step": 0}
step = 0

def run_eval(epoch):
    tl, _, _ = evaluate_loader(train_loader, model, device, max_batches=10)
    vl, vp, vlab = evaluate_loader(val_loader, model, device)
    steps.append(step); train_losses.append(tl); val_losses.append(vl)
    if vl < best["val_loss"]:
        best["val_loss"], best["step"] = vl, step
        best["state"] = {n: p.detach().clone() for n, p in model.named_parameters() if p.requires_grad}
    print(f"Ep {epoch} step {step}: train {tl:.3f} | val {vl:.3f} | val acc {accuracy_score(vlab, vp):.3f} | val macroF1 {f1_score(vlab, vp, average='macro', zero_division=0):.3f}")

t0 = time.time()
for epoch in range(1, EPOCHS + 1):
    for x, y in train_loader:
        model.train()
        optimizer.zero_grad()
        loss = calc_loss_batch(x, y, model, device)
        loss.backward()
        optimizer.step()
        step += 1
        if step % EVAL_FREQ == 0:
            run_eval(epoch)
    if step % EVAL_FREQ != 0:
        run_eval(epoch)   # end-of-epoch evaluation
train_minutes = (time.time() - t0) / 60
print(f"Done in {train_minutes:.1f} min")

# restore best-validation weights
with torch.no_grad():
    for n, p in model.named_parameters():
        if n in best["state"]:
            p.copy_(best["state"][n])
print(f"Restored best weights from step {best['step']} (val loss {best['val_loss']:.4f})")

# %%
# CELL 11: loss curves
plt.figure(figsize=(7, 4))
plt.plot(steps, train_losses, label="Train loss")
plt.plot(steps, val_losses, label="Validation loss")
plt.xlabel("Training step"); plt.ylabel("Cross-entropy loss")
plt.legend(); plt.grid(alpha=.3); plt.title("SciCite fine-tuning loss"); plt.tight_layout()
plt.savefig("loss_curve.png", dpi=200); plt.show()

# %%
# CELL 12: final TEST evaluation: accuracy, F1, confusion matrix (run on test only for final numbers)
t_loss, t_pred, t_lab = evaluate_loader(test_loader, model, device)
after = metrics(t_pred, t_lab)

print("BEFORE:", baseline)
print("AFTER :", after, "| test loss: %.4f" % t_loss)
print("Majority-class accuracy on test: %.4f" % (np.bincount(t_lab).max() / len(t_lab)))
print(classification_report(t_lab, t_pred, target_names=LABELS, digits=3, zero_division=0))

cm = confusion_matrix(t_lab, t_pred, labels=[0, 1, 2])
fig, ax = plt.subplots(figsize=(4.8, 4.2))
ax.imshow(cm, cmap="Blues")
ax.set_xticks(range(3)); ax.set_xticklabels(LABELS); ax.set_yticks(range(3)); ax.set_yticklabels(LABELS)
ax.set_xlabel("Predicted"); ax.set_ylabel("True"); ax.set_title("Confusion matrix (test)")
for i in range(3):
    for j in range(3):
        ax.text(j, i, cm[i, j], ha="center", va="center", color="white" if cm[i, j] > cm.max() / 2 else "black")
plt.tight_layout(); plt.savefig("confusion_matrix.png", dpi=200); plt.show()

# %%
# CELL 13: classify(text) + real error examples
@torch.no_grad()
def classify(text, mdl=None, max_length=MAX_LEN):
    mdl = model if mdl is None else mdl
    mdl.eval()
    ids = encode(text)[:max_length]
    ids = ids + [PAD_ID] * (max_length - len(ids))
    x = torch.tensor(ids).unsqueeze(0).to(device)
    probs = torch.softmax(get_logits(mdl, x), dim=-1)[0].cpu().numpy()
    return LABELS[int(probs.argmax())], {l: float(p) for l, p in zip(LABELS, probs)}

# These two are my own illustrative sentences, NOT dataset entries.
print(classify("We used the Adam optimizer with a learning rate of 0.001 as in previous work."))
print(classify("Our approach achieved 92% accuracy, outperforming the baseline."))

# Misclassified TEST sentences for the report's "error analysis"
wrong = [i for i in range(len(t_lab)) if t_pred[i] != t_lab[i]]
random.Random(SEED).shuffle(wrong)
print("\nTotal test errors:", len(wrong), "of", len(t_lab))
for i in wrong[:10]:
    print(f"TRUE={LABELS[t_lab[i]]:10s} PRED={LABELS[t_pred[i]]:10s} | {test_rows[i][0][:200]}")

# %%
# CELL 14: save, reload from scratch, verify
SAVE_PATH = "scicite_gpt2_classifier.pt"
torch.save({"state_dict": model.state_dict(), "labels": LABELS, "max_len": MAX_LEN,
            "config": GPT_CONFIG_124M}, SAVE_PATH)
print("Saved locally:", SAVE_PATH, f"({os.path.getsize(SAVE_PATH) / 1e6:.0f} MB)")

# Optional: copy to Google Drive so a disconnect does not lose it
SAVE_TO_DRIVE = True
if SAVE_TO_DRIVE:
    try:
        from google.colab import drive
        drive.mount('/content/drive')
        get_ipython().system('cp scicite_gpt2_classifier.pt /content/drive/MyDrive/')
        print("Backed up to Google Drive (MyDrive).")
    except Exception as e:
        print("Drive backup failed, continuing:", e)

# Rebuild an empty model; no pretrained download needed, everything is in the checkpoint
model2 = GPTModel(GPT_CONFIG_124M)
model2.out_head = nn.Linear(GPT_CONFIG_124M["emb_dim"], 3)
ckpt = torch.load(SAVE_PATH, map_location=device)
model2.load_state_dict(ckpt["state_dict"])
model2.to(device).eval()

test_texts = [r[0] for r in test_rows[:200]]
same = sum(classify(t, model)[0] == classify(t, model2)[0] for t in test_texts)
print(f"Reloaded model agrees with in-memory model on {same}/200 predictions")
_, p2, l2 = evaluate_loader(test_loader, model2, device)
print("Reloaded test metrics:", metrics(p2, l2), "(should equal AFTER in Cell 12)")

# %%
# CELL 15: Gradio interface (uses the RELOADED model2)
import gradio as gr

def predict_ui(text):
    if not text or not text.strip():
        return {l: 0.0 for l in LABELS}
    return classify(text, model2)[1]

demo = gr.Interface(
    fn=predict_ui,
    inputs=gr.Textbox(lines=4, label="Citation sentence"),
    outputs=gr.Label(num_top_classes=3, label="Predicted citation intent"),
    title="Citation Intent Classifier (GPT-2 fine-tuned on SciCite)",
    examples=[["We follow the preprocessing pipeline described in [12]."],
              ["Deep learning has transformed natural language processing [3]."],
              ["Our F1 score is higher than that reported by [7]."]])
demo.launch(debug=False)

# %%
# CELL 16: versions, requirements.txt and results summary (for report Appendix A.1 and Tables 4.1 / 3.2)
import sys, importlib.metadata as md
print("Python", sys.version.split()[0])
pins = []
for pkg in ["torch", "tiktoken", "numpy", "scikit-learn", "matplotlib", "gradio", "tensorflow", "tqdm"]:
    try:
        v = md.version(pkg); print(f"{pkg}=={v}"); pins.append(f"{pkg}=={v}")
    except Exception:
        print(pkg, "not found")
with open("requirements.txt", "w") as f:
    f.write("\n".join(pins) + "\n")

summary = {
    "seed": SEED, "device": str(device), "gpu": torch.cuda.get_device_name(0) if torch.cuda.is_available() else None,
    "sizes": {"train": len(train_rows), "val": len(val_rows), "test": len(test_rows)},
    "max_len": MAX_LEN, "batch_size": BATCH_SIZE, "epochs": EPOCHS, "lr": LR,
    "unfrozen_blocks": N_UNFREEZE_BLOCKS, "total_params": total, "trainable_params": trainable,
    "train_minutes": round(train_minutes, 2), "best_step": best["step"],
    "baseline_test": baseline, "baseline_test_loss": round(b_loss, 4),
    "after_test": after, "after_test_loss": round(t_loss, 4),
    "majority_acc_test": round(float(np.bincount(t_lab).max() / len(t_lab)), 4),
}
with open("results_summary.json", "w") as f:
    json.dump(summary, f, indent=2)
print(json.dumps(summary, indent=2))

# %%
# CELL 17 (Task 3, step 1): upload ONE real paper's PDF and list its citation sentences
get_ipython().system('pip install -q pypdf')
import re
from google.colab import files
from pypdf import PdfReader

up = files.upload()                      # click "Choose Files" and select the paper's PDF
fname = list(up.keys())[0]
raw_text = " ".join((p.extract_text() or "") for p in PdfReader(fname).pages)

def split_candidates(text):
    text = re.sub(r"Authorized licensed use limited to.*?Restrictions apply\.", " ", text, flags=re.S)
    text = re.sub(r"-\s*\n\s*", "", text)
    text = " ".join(text.split())
    text = text.split("REFERENCES")[0]
    sents = re.split(r"(?<=[.!?])\s+(?=[A-Z])", text)
    return [s.strip() for s in sents if re.search(r"\[\d+", s) and 12 <= len(s.split()) <= 60]

cands = split_candidates(raw_text)
print(len(cands), "candidate sentences\n")
for i, s in enumerate(cands):
    print(i, "|", s, "\n")

# %%
# CELL 18 (Task 3, step 2): choose 10 sentences by index, give YOUR OWN label, and classify them
# Read the list printed by Cell 17 first. Change indexes/labels to your own judgment.
# Labels: background / method / result. Skip any index that is a heading fragment or not a citation sentence.
CHOSEN = {
    1: "background",
    2: "background",
    4: "background",
    5: "background",
    6: "background",
    7: "result",
    8: "background",
    9: "background",
    10: "method",
    12: "method",
}
assert len(CHOSEN) == 10, f"Pick exactly 10, you picked {len(CHOSEN)}"
assert max(CHOSEN) < len(cands), "An index is larger than the candidate list; check Cell 17 output"

PAPER = ('O. Al-Refai, I. Shahbaz and E. Hammad, "Composable Trust in Agentic AI: Bridging Architectural '
         'Capability and System-Level Assurance," 2026 56th IEEE/IFIP DSN Workshops (DSN-W), pp. 17-20, '
         'doi: 10.1109/DSN-W70714.2026.00023.')

def citation_intent(sentence):
    """Statement-required function: returns (label, probabilities) = background / method / result."""
    return classify(sentence, model2)

print("Paper:", PAPER, "\n")
agree = 0
for i, exp in CHOSEN.items():
    assert exp in LABELS, f"Bad label '{exp}'"
    s = cands[i]
    assert len(s.split()) >= 12, f"Index {i} is too short, choose another"
    label, probs = citation_intent(s)
    agree += (label == exp)
    print(f"expected={exp:10s} predicted={label:10s} conf={max(probs.values()):.2f} words={len(s.split()):3d} | {s[:80]}")
print(f"\nAgreement with my own labels: {agree}/{len(CHOSEN)}")
```
